# Experiment Analysis

This notebook demonstrates how to load and analyze the results of an experiment.

```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join('..')))
from src.analysis import load_results, plot_training_curve, generate_report

# --- Parameters ---
# Replace with the name of the experiment you want to analyze
# The 'test_experiment' will be created when you run the tests.
experiment_name = "test_experiment" 
# ------------------

exp_dir = os.path.join('..', 'results', experiment_name)

# You can generate the full report programmatically
generate_report(exp_dir)

# Or, you can load the data and plot interactively
try:
    results, config = load_results(exp_dir)
    print("\n--- Interactive Plot ---")
    plot_training_curve(results)
except FileNotFoundError:
    print(f"\nExperiment '{experiment_name}' not found. \nPlease run the tests or a training session first.")
```
